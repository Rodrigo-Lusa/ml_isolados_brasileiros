```python
import argos
from argos import ArgosDataset
import pandas as pd
import numpy as np
```

## Importando dados

```python
metadata = pd.read_csv("../data/processed/metadata_filtrada.csv")
metadata.head()
```

```python
ids_finais = metadata["sample"].tolist()
dataset = ArgosDataset.from_parquet("../data/argos_project/parquet_data", samples=ids_finais)
dataset.resistome.head()
```

```python
vfdb = pd.read_csv('../data/raw/vfdb_all_hits.csv')

vfdb.head()
```

## Fazendo as `features`

### Ajustando metadata

```python
# Reindexando o `metadata` pra eliminar erro futuro
metadata = metadata.set_index("sample")
metadata.head(n=3)
```

### Matrizes

```python
def matriz_por_contig_type(resistome, contig_type) -> pd.DataFrame:
    hits = resistome[resistome['contig_type'] == contig_type]
    matriz = pd.crosstab(hits['sample_id'], hits['gene_clean']).clip(upper=1)
    return matriz.reindex(metadata.index, fill_value= 0) # pra genomas sem hits == 0
```

```python
matriz_cromossomal = matriz_por_contig_type(dataset.resistome, "chromosome")
matriz_cromossomal.head(n=3)
```

```python
matriz_plasmidial = matriz_por_contig_type(dataset.resistome, "plasmid")
matriz_plasmidial.head(n=3)
```

### n_genes e mdr_score

```python
n_genes_cromossomal = matriz_cromossomal.sum(axis=1).rename("n_genes_cromossomal")
n_genes_plasmidial = matriz_plasmidial.sum(axis=1).rename("n_genes_plasmidial")

n_genes_plasmidial.head()
```

```python
# Método do argos que agrupa por classe de droga
argos.groupby_genes(matriz_cromossomal, dataset.annotations, by='drug_class_compacted', agg='max').head(n=3)
```

```python
mdr_score_cromossomal = argos.groupby_genes(
    matriz_cromossomal, dataset.annotations, by="drug_class_compacted", agg="max"
).sum(axis=1).rename("mdr_score_cromossomal")

mdr_score_plasmidial = argos.groupby_genes(
    matriz_plasmidial, dataset.annotations, by="drug_class_compacted", agg="max"
).sum(axis=1).rename("mdr_score_plasmidial")

mdr_score_plasmidial.head()
```

### Tipo de contig

```python
contagem_tipo_contig = pd.crosstab(dataset.contigs["sample_id"], dataset.contigs["contig_type"])
contagem_tipo_contig.head()
```

```python
percentual_contig = contagem_tipo_contig.div(contagem_tipo_contig.sum(axis=1), axis=0) * 100
percentual_contig = percentual_contig.reindex(metadata.index, fill_value=0)
percentual_contig.head()
```

```python
pct_contigs_cromossomal = percentual_contig.get("chromosome", 0).rename("pct_contigs_cromossomal")
pct_contigs_plasmidial = percentual_contig.get("plasmid", 0).rename("pct_contigs_plasmidial")

pct_contigs_plasmidial.head()
```

### Contigs plasmidiais

```python
contigs_plasmid = dataset.contigs[dataset.contigs["contig_type"] == "plasmid"].copy()
contigs_plasmid.head(n=3)
```

```python
contigs_plasmid.value_counts('topology')
```

```python
topology_pct = (
    pd.crosstab(contigs_plasmid["sample_id"], contigs_plasmid["topology"], normalize="index") * 100
).reindex(metadata.index, fill_value=0)

topology_pct.columns = [
    f"pct_plasmid_topology_{c.lower().replace(' ', '_')}" for c in topology_pct.columns
]

# pra cada genoma, dos plasmídeos que ele tem, quanto % é circular fechado (DTR/ITR) ou fragmento linear sem repetição terminal
topology_pct.head()
```

```python
contigs_plasmid["tem_amr"] = contigs_plasmid["amr"] > 0
contigs_plasmid["tem_conjugacao"] = contigs_plasmid["n_conjugation_genes"] > 0
contigs_plasmid["conjugacao_e_amr"] = contigs_plasmid["tem_amr"] & contigs_plasmid["tem_conjugacao"]

resumo_plasmid_contig = contigs_plasmid.groupby("sample_id").agg(
    n_contigs_plasmidiais=("seq_name", "count"),
    n_contigs_plasmid_amr=("tem_amr", "sum"),
    n_contigs_plasmid_conj_amr=("conjugacao_e_amr", "sum"),
).reindex(metadata.index, fill_value=0)

resumo_plasmid_contig.head()
```

```python
# Transforma em %
pct_plasmidial_amr = (
    100 * resumo_plasmid_contig["n_contigs_plasmid_amr"] / resumo_plasmid_contig["n_contigs_plasmidiais"]
).fillna(0).rename("pct_plasmidial_amr")

pct_plasmidial_conjugacao_amr = (
    100 * resumo_plasmid_contig["n_contigs_plasmid_conj_amr"] / resumo_plasmid_contig["n_contigs_plasmidiais"]
).fillna(0).rename("pct_plasmidial_conjugacao_amr")

pct_plasmidial_conjugacao_amr.head()
```

```python
genes_conjugativos_por_amostra = (
    contigs_plasmid[contigs_plasmid["tem_conjugacao"]][["sample_id", "amr_list"]]
    .explode("amr_list").dropna(subset=["amr_list"])
    .groupby("sample_id")["amr_list"].nunique()
)
n_genes_plasmidial_conjugativo = genes_conjugativos_por_amostra.reindex(
    metadata.index, fill_value=0
).rename("n_genes_plasmidial_conjugativo")

# número de genes AMR distintos presentes em contigs plasmidiais que também carregam genes de conjugação especificamente (n_conjugation_genes > 0)
n_genes_plasmidial_conjugativo.head()
```

### Matriz de genes por drug class (em contigs plasmidiais)

```python
matriz_drugclass_plasmidial = argos.groupby_genes(
    matriz_plasmidial, dataset.annotations, by="drug_class_compacted", agg="sum"
)
matriz_drugclass_plasmidial.columns = [
    f"n_plasmidial_{c.lower().replace('/', '_').replace(' ', '_')}" for c in matriz_drugclass_plasmidial.columns
]

matriz_drugclass_plasmidial.head()
```

### Matriz de classes de genes de virulencia (vfdb - matriz preparada anteriormente)

```python
vfdb.head(n=2)
```

```python
vfdb["sample_id"] = vfdb["sample"].str.extract(r"(GC[AF]_\d+\.\d+)")

print(vfdb["sample_id"].isna().sum())
vfdb.head(n=2)
```

```python
def matriz_vfdb_por_contig_type(vfdb, contig_type_map, contig_type):
    hits = vfdb[vfdb["SEQUENCE"].map(contig_type_map) == contig_type]
    matriz = pd.crosstab(hits["sample_id"], hits["categoria"])   # <- era hits["sample"]
    matriz = matriz.reindex(metadata.index, fill_value=0)
    matriz.columns = [f"vfdb_{contig_type}_{c.lower().replace('/', '_').replace(' ', '_')}" for c in matriz.columns]
    return matriz
```

```python
contig_type_map = dataset.contigs.set_index("seq_name")["contig_type"]

vfdb_cromossomal = matriz_vfdb_por_contig_type(vfdb, contig_type_map, "chromosome")
vfdb_plasmidial = matriz_vfdb_por_contig_type(vfdb, contig_type_map, "plasmid")

print(vfdb_plasmidial.sum().sum())
vfdb_plasmidial.loc["GCA_009651075.1"]
```

```python
virulence_score_cromossomal = vfdb_cromossomal.clip(upper=1).sum(axis=1).rename("virulence_score_cromossomal")
virulence_score_plasmidial = vfdb_plasmidial.clip(upper=1).sum(axis=1).rename("virulence_score_plasmidial")

virulence_score_plasmidial.loc["GCA_009651075.1"]  # tem 1 ser 4
```

### Verificando provirus - bacteriófago que está integrado no genoma

```python
provirus_genes = dataset.genomad_genes[
    dataset.genomad_genes["gene_id"].str.contains(r"\|provirus_\d+_")
].copy()

print(provirus_genes.shape)
provirus_genes.head()
```

```python
# extrai o contig hospedeiro (antes do "|") de cada gene de provírus
provirus_genes["host_contig"] = provirus_genes["seq_name"].str.split("|").str[0]

provirus_genes.head()
```

```python
host_contig_type = dataset.contigs.set_index("seq_name")["contig_type"]
provirus_genes["host_contig_type"] = provirus_genes["host_contig"].map(host_contig_type)

provirus_genes.head()
```

```python
tem_provirus_em_plasmidio = (
    (provirus_genes["host_contig_type"] == "plasmid")
    .groupby(provirus_genes["sample_id"]).any()
    .reindex(metadata.index, fill_value=False).astype(int)
    .rename("tem_provirus_em_plasmidio")
)

tem_provirus_em_plasmidio.head()
```

```python
tem_amr_provirus = (
    provirus_genes.groupby("sample_id")["has_amr"].any()
    .reindex(metadata.index, fill_value=False).astype(int)
    .rename("tem_amr_em_provirus")
)

tem_amr_provirus.head()
```

```python
print(tem_amr_provirus.sum(), "genomas com AMR em provírus")
print(tem_provirus_em_plasmidio.sum(), "genomas com provírus em plasmídeo")
```

```python
n_proviruses = (
    provirus_genes.groupby("sample_id")["seq_name"].nunique()
    .reindex(metadata.index, fill_value=0)
    .rename("n_proviruses")
)

n_proviruses.head()
```

## Fazendo a tabela final

```python
raw_features = pd.concat([
    # resistoma — cromossomal vs plasmidial
    n_genes_cromossomal,
    n_genes_plasmidial,
    mdr_score_cromossomal,
    mdr_score_plasmidial,
    matriz_drugclass_plasmidial,          # n_plasmidial_<classe_de_droga>

    # estrutura de contig
    pct_contigs_cromossomal,
    pct_contigs_plasmidial,
    topology_pct,                          # pct_plasmid_topology_{dtr,itr,no_terminal_repeats}

    # mobilidade / conjugação
    pct_plasmidial_amr,
    pct_plasmidial_conjugacao_amr,
    n_genes_plasmidial_conjugativo,

    # virulência (VFDB)
    vfdb_plasmidial,                       
    virulence_score_plasmidial,
    virulence_score_cromossomal,

    # provírus
    n_proviruses,
    tem_provirus_em_plasmidio,
    tem_amr_provirus,
], axis=1)

print(raw_features.shape)
raw_features.head()
```

```python
raw_features.to_csv('../data/raw/raw_features.csv')
```

## Exportando a lista de contigs cromossomais e plasmidiais de cada genoma pra calcular %GC

```python
contigs_por_tipo = (
    dataset.contigs
    .groupby(["sample_id", "contig_type"])["seq_name"]
    .apply(list)
    .unstack("contig_type")
    .reindex(metadata.index)
)

contigs_cromossomais = contigs_por_tipo["chromosome"].apply(lambda x: x if isinstance(x, list) else [])
contigs_plasmidiais = contigs_por_tipo["plasmid"].apply(lambda x: x if isinstance(x, list) else [])

contig_lists = pd.DataFrame({
    "contigs_cromossomais": contigs_cromossomais,
    "contigs_plasmidiais": contigs_plasmidiais,
})
contig_lists.head()
```

```python
contig_lists.to_csv("../data/raw/contig_lists.csv")
```

## Adicionando %GC e salvando

```python
gc = pd.read_csv("../data/raw/gc_stats.txt", sep=r"\s+")

gc.head()
```

```python
gc["sample_id"] = gc["file"].str.extract(r"(GC[AF]_\d+\.\d+)")
gc["contig_type"] = gc["file"].str.extract(r"_(chromosome|plasmid)\.fna$")

gc.head()
```

```python
gc_wide = gc.pivot(index="sample_id", columns="contig_type", values="GC(%)")

gc_wide.head()
```

```python
gc_wide.columns = ["gc_cromossomal", "gc_plasmidial"]
gc_wide["gc_diff_plasmid_cromossomo"] = gc_wide["gc_plasmidial"] - gc_wide["gc_cromossomal"]

gc_wide.head()
```

```python
raw_features = pd.read_csv("../data/raw/raw_features.csv", index_col=0)

raw_features = raw_features.join(gc_wide, how="left")

print(raw_features.shape)
```

```python
raw_features.head()
```

```python
sem_plasmidio = raw_features["pct_contigs_plasmidial"] == 0  

raw_features.loc[sem_plasmidio, "gc_plasmidial"] = np.nan
raw_features.loc[sem_plasmidio, "gc_diff_plasmid_cromossomo"] = np.nan
```
